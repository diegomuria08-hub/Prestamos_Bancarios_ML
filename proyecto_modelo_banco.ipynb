# Proyecto de Machine Learning: Clasificador de Préstamos Bancarios con XGBoost
Este cuaderno contiene el flujo de entrenamiento y exportación automatizada del modelo hacia Google Drive.

```python
import os
import pandas as pd
from xgboost import XGBClassifier
from google.colab import drive

# 1. CONEXIÓN A GOOGLE DRIVE
drive.mount('/content/drive')
ruta_drive = '/content/drive/My Drive/Cursos de IA'

if not os.path.exists(ruta_drive):
    os.makedirs(ruta_drive)

# 2. DATASET SIMULADO
datos_banco = {
    'Ingresos': [45000, 32000, 85000, 120000, 25000, 55000, 95000, 30000, 110000, 40000],
    'Score_Credito': [650, 580, 710, 790, 500, 680, 740, 600, 810, 620],
    'Edad': [34, 28, 45, 50, 22, 39, 41, 26, 48, 31],
    'Pago_Prestamo': [1, 0, 1, 1, 0, 1, 1, 0, 1, 1]
}
df = pd.DataFrame(datos_banco)
X_entreno = df[['Ingresos', 'Score_Credito', 'Edad']]
y_entreno = df['Pago_Prestamo']

# 3. ENTRENAMIENTO
modelo_banco = XGBClassifier(n_estimators=50, learning_rate=0.1, max_depth=2, random_state=42)
modelo_banco.fit(X_entreno, y_entreno)

# 4. GUARDAR MODELO
ruta_json_final = os.path.join(ruta_drive, 'modelo_banco_xgb.json')
modelo_banco.save_model(ruta_json_final)
print(f'✅ ¡Modelo guardado en: {ruta_json_final}!')
```
